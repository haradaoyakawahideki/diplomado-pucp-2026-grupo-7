# Parte 2 – Tuplas

Crearemos una tupla que contiene edades y realizaremos diferentes operaciones, como mostrar sus elementos, encontrar el valor máximo y mínimo, contar sus elementos e intentar modificar uno de ellos.

```python
# Crear la tupla
ages = (20, 25, 22, 30, 28)
```

## 1. Mostrar la tupla

Primero mostramos todos los elementos que contiene la tupla.

```python
print("Tupla:", ages)
```

## 2. Mostrar el primer y último elemento

Para acceder a los elementos de una tupla utilizamos su posición.

El primer elemento tiene el índice `0` y para acceder al último elemento podemos utilizar el índice `-1`.

```python
print("Primer elemento:", ages[0])
print("Último elemento:", ages[-1])
```

## 3. Valor máximo, mínimo y número de elementos

Para obtener información de la tupla utilizaremos las siguientes funciones:

- `max()` permite obtener el valor máximo.
- `min()` permite obtener el valor mínimo.
- `len()` permite conocer la cantidad de elementos.

```python
print("Valor máximo:", max(ages))
print("Valor mínimo:", min(ages))
print("Número de elementos:", len(ages))
```

## 4. Intentar modificar un elemento

Ahora intentaremos modificar el primer elemento de la tupla.

La instrucción sería:

`ages[0] = 21`

Al ejecutar esta instrucción, Python genera un error porque las tuplas no permiten modificar sus elementos una vez que han sido creadas.

Para que el notebook final pueda ejecutarse completamente sin errores, dejaremos esta instrucción comentada.

### Intento de modificar el primer elemento de la tupla
 ages[0] = 21

## 5. ¿Por qué no se puede modificar una tupla?

Python no permite modificar los elementos de una tupla porque las tuplas son **inmutables**.

Esto significa que, una vez creada una tupla, sus elementos no pueden ser modificados, agregados o eliminados.
