```python
population = {
    "Lima": 10092000,
    "Arequipa": 1008000,
    "Trujillo": 919000
}

print(population)
```

Un diccionario guarda pares de llave y valor. Aquí la llave es el nombre de la ciudad y el valor es su población, así que para encontrar un dato uso el nombre y no un número de posición como en las listas.

```python
print(population.keys())
```

keys() devuelve solo las llaves, es decir, los nombres de las ciudades sin las poblaciones.

```python
print(population.get("Lima"))
```

get() busca el valor usando la llave. También podría usar population["Lima"], pero con get(), si la llave no existe, el programa no da error y devuelve None.

```python
population["Cusco"] = 428000
print(population)
```

Para agregar un dato nuevo escribo la llave nueva y le asigno su valor, y se añade al final del diccionario.

```python
population.pop("Trujillo")
print(population)
```

pop() elimina la llave que le indico y devuelve su valor. Al imprimir se ve que Trujillo ya no está en el diccionario.

```python
print(population)
```

Este es el diccionario final: quedó con Lima, Arequipa y Cusco.
