# Assignment 1 – Parte 3: Diccionarios

Ejercicios con el diccionario `population`.

## Crear el diccionario

```python
population = {
    "Lima": 10092000,
    "Arequipa": 1008000,
    "Trujillo": 919000
}
```

## 1. Imprimir el diccionario completo

```python
print("Diccionario completo:", population)
```

## 2. Mostrar solo las claves con `keys()`

```python
print("Claves:", population.keys())
```

## 3. Obtener la población de "Lima" con `get()`

```python
print("Población de Lima:", population.get("Lima"))
```

## 4. Agregar "Cusco": 428000

```python
population["Cusco"] = 428000
print("Después de agregar Cusco:", population)
```

## 5. Eliminar "Trujillo" con `pop()`

```python
removed = population.pop("Trujillo")
print("Población eliminada de Trujillo:", removed)
```

## 6. Imprimir el diccionario final

```python
print("Diccionario final:", population)
```
