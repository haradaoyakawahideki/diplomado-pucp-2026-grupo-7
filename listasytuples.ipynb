# PREGUNTA 1: lists.ipynb
## Crear la lista de notas

```python

grades = [15, 18, 12, 20, 16]

# 1. Mostrar la lista original
print("Original list:", grades)
# 2. Agregar la nota 17
grades.append(17)

print("List after append:", grades)
# 3. Ordenar las notas de menor a mayor
grades.sort()

print("Sorted list:", grades)
# 4. Mostrar la nota máxima, mínima y cantidad de notas
print("Highest grade:", max(grades))
print("Lowest grade:", min(grades))
print("Number of grades:", len(grades))
# 5. Mostrar la lista final
print("Final list:", grades)
```
